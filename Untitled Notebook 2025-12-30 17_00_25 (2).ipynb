```python
#import libraries 
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
```

```python
spark = SparkSession. builder. appName("NBA_Player_Statistics_ETL") .getOrCreate()
```

```python
#load the data set and dsiplay to have a better understanding
player_data = spark.read.csv(
    '/Volumes/workspace/default/my_volume/PlayerStatistics.csv',
    header=True,
    inferSchema=True
)
```

```python
#inspect the data
player_data.show(5)
player_data.printSchema ()
```

```python
#clean the data
#removed rows with duplicate or missing values
clean_data = player_data.dropna(
    subset=["personId", "gameId", "gameDateTimeEst", "playerteamName", "opponentteamName" ]
)
clean_data = clean_data.dropDuplicates(
    ["personId", "gameId"]
)
```

```python
#filled in missing values in these fields with 0
clean_data = clean_data.fillna({
 "points": 0,
 "assists": 0,
 "reboundsTotal": 0
})
#converted the following columns to integer to ease analysis
clean_data = clean_data \
 .withColumn("points", col("points").cast("int")) \
 .withColumn("assists", col("assists").cast("int")) \
 .withColumn("reboundsTotal", col("reboundsTotal").cast("int"))
```

```python
#higlighted the top 10 scorer by points
topscorer = clean_data.groupBy("firstName", "lastName") \
 .sum("points") \
 .orderBy(col("sum(points)").desc())

topscorer.show(10)
```

```python
# found the average of the wins of each team to calculate win rate
from pyspark.sql.functions import avg
win_rate = clean_data.groupBy("playerteamName") \
 .agg(avg("win").alias("win_rate")) \
 .orderBy("win_rate", ascending=False)

win_rate.show()
```

```python
#calculate average point per player in each team
avgptsperplayerinteam = clean_data.groupBy("playerteamName") \
 .agg(avg("points").alias("avgpointsper_player")) \
 .orderBy(col("avgpointsper_player").desc())

avgptsperplayerinteam.show(10)
```

```python
#saved the clean data as parquet
clean_data.write \
    .mode("overwrite") \
    .parquet("/Volumes/workspace/default/my_volume/cleaned_player_stats")
```
